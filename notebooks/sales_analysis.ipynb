# Sales Insights — Exploratory Analysis


```python
import pandas as pd
df = pd.read_csv('../data/processed/sales_transactions_clean.csv', parse_dates=['order_date'])
df.head()
```

```python
pd.Series({
 'Total Sales': df.sales_amount.sum(),
 'Total Profit': df.profit.sum(),
 'Orders': df.order_id.nunique(),
 'Units Sold': df.quantity.sum(),
 'Profit Margin': df.profit.sum()/df.sales_amount.sum(),
 'AOV': df.sales_amount.sum()/df.order_id.nunique()
})
```

```python
df.groupby('category').agg(sales=('sales_amount','sum'), profit=('profit','sum')).sort_values('sales',ascending=False)
```

```python
monthly=df.groupby('year_month',as_index=False).agg(sales=('sales_amount','sum'))
monthly.plot(x='year_month',y='sales',figsize=(11,4),marker='o',title='Monthly Sales Trend')
```
